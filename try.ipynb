```python
import pandas as pd

# df2 columns assumed: user_id, track_id, play_count

song_stats = (
    df2.groupby('track_id')
       .agg(
           total_playcount=('playcount', 'sum'),
           n_users=('user_id', 'nunique')
       )
       .reset_index()
)

song_stats['freq'] = song_stats['total_playcount'] / song_stats['n_users']

song_stats = song_stats.sort_values('total_playcount', ascending=False)

song_stats = song_stats[song_stats['n_users'] > 5]
song_stats
```

```python
df = pd.merge(df1, song_stats, on='track_id', how='inner')
print(df.shape)
df
```

```python
df.isnull().sum()
```

```python
df["tags"] = df["tags"].fillna("non")
df.isnull().sum()
```

```python
df = df.drop(columns=["total_playcount","n_users","freq"])
df.info()
```

```python
df.info()
```

```python
from sklearn.preprocessing import MinMaxScaler

numeric_cols = df.select_dtypes(include="number").columns

scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])
```

```python
df.to_csv("initial_data.csv")
```
